```python
import sqlite3

def addMeal(meal_plan, date, meal, recipe_id):
    connection = sqlite3.connect("database/RecipeBook.db")
    cursor = connection.cursor()
    result = cursor.execute("SELECT * FROM Recipes WHERE ID = ?", [recipe_id])
    recipe = result.fetchone()
    
    meal_plan[date][meal] = recipe[1]
```

```python
from datetime import datetime, timedelta, date
addMeal(meal_plan, date(2026, 9, 20), "breakfast", 1)
```

```python
def mealCounter(meal_plan):
    meal_counts = {}
    
    for day, meals in meal_plan.items():
        for time, recipe in meals.items():
            if recipe in meal_counts:
                meal_counts[recipe] += 1
            else:
                meal_counts[recipe] = 1

    return meal_counts
```

```python
print(mealCounter(meal_plan))
```
