```python
from PyQt6.QtWidgets import QApplication, QMainWindow, QLabel, QMessageBox
from PyQt6 import QtWidgets
import sys
```

```python
class Ui_MainWindow(QMainWindow):
    def __init__(self):
        super(Ui_MainWindow, self).__init__()
        self.setGeometry(50, 50, 600, 400)
        self.setWindowTitle("Hey girl")

        self.setupUI()

        self.button1.clicked.connect(self.show_dialogue)

    def show_dialogue(self):
        dialogue = QMessageBox(self.MainWindow)
        dialogue.setText("I'm not keeping track")
        dialogue.setTitle('poopy')
        dialogue.exec()

    def setupUI(self):
        self.text = QLabel(self)
        self.text.setText("what's good")
        self.text.move(250, 175)

        self.button1 = QtWidgets.QPushButton(self)
        self.button1.setText("push me")
        self.button1.clicked.connect(self.on_click)

    def on_click(self):
        self.text.setText("why though")
```

```python
def main():
    app = QApplication(sys.argv)
    window = Ui_MainWindow()

    window.show()
    sys.exit(app.exec())
```

```python
main()
```
